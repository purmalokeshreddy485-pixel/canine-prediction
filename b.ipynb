```python
import tensorflow as tf
import cv2
import numpy as np
```

```python
image=cv2.imread("b.jpg")
image=cv2.resize(image,(256,256))
image=image/255.0
model=tf.keras.models.load_model("canine_breed.h5")
```

```python
a=model.predict(np.array([image]))
```

```python
b=np.argmax(a)
```

```python
classnames=["Bedlington_terrier","Bernese_mountain_dog","Dandie_Dinmont","Gordon_setter","Ibizan_hound","Norwegian_elkhound"]
print("the predicted tumor type is: "+classnames[b])
```
